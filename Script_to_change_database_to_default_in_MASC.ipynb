```python
#----------------------Create User ---------------------------#
# API endpoint
#url = "https://useip.sce.manh.com/organization/api/organization/user/save"
url = "https://useiv2.sce.manh.com/organization/api/organization/user/save"

# Headers
headers = {
    "Content-Type": "application/json",
    "selectedOrganization": "LF9",
    "selectedLocation": "LF9",
    "Authorization": f"Bearer {token}"
}
for index, row in df.iterrows():
    # User data payload
    user_data = {
       "UserId": row["User ID"],
       "IdentityType": {
            "IdentityTypeId": "Default"
        }
    }

    # Send POST request
    try:
        response = requests.post(url, headers=headers, json=user_data)
        if response.status_code == 200:
            print(f"✅ User {row["User ID"]} changed successfully!")
        else:
            print(f"❌ Failed to change user {row["User ID"]}. Status code: {response.status_code}")
            print("Response:", response.text)
    except Exception as e:
        print(f"⚠️ Error changing user {row["User ID"]}: {e}")
```

```python
#----------------------Create Facility (Run only if it is L2 Org) ------------#
# API endpoint
url = "https://useip.sce.manh.com/organization/api/organization/user/save"

# Headers
headers = {
    "Content-Type": "application/json",
    "selectedOrganization": "LF9",
    "selectedLocation": "LF9",
    "Authorization": f"Bearer {token}"
}
for index, row in df.iterrows():
    # User data payload
    user_data = {
        "UserId": row["UserId"],
        "UserLocation": [
            {
                "LocationId": row["Org Id"]
            }
        ]
    }

    # Send POST request
    try:
        response = requests.post(url, headers=headers, json=user_data)
        if response.status_code == 200:
            print(f"✅ Facility {row['Org Id']} created successfully!")
        else:
            print(f"❌ Failed to create facility {row['Org Id']}. Status code: {response.status_code}")
            print("Response:", response.text)
    except Exception as e:
        print(f"⚠️ Error creating facility  {row['Org Id']}: {e}")
```
